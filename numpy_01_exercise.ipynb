### **What is difference arnage and range functions.**

* range() : 
    1. Output Types :  Immutable sequences of integers.
    2. Data Types : Intgers Only 
    3. Usage : Loops, Iteration
    4. Example : range(start,stop, step)

* arange():
    1. Output Types: Numpy array (can be int or float)
    2. Data Types: Integers or floats
    3. Numberic Computation array generation
    4. Example : np.arange(start,stop,step)

## **List VS Numpy Array**

* List : 
    1. Contain Different data Typs like int, float, complex
    2.  Objects , Require Higher memory usage
    3. Sloawer interpreted per item
    4. No built in math operations

* NumPy Array :
    1. Homogeneity Maintain, Single Data types
    2. Memory less Required compare to list, Contiguous
    3. Speed : Vectorized, Much Faster
    4. Rich math/ statistics toolkits


```python
# Element wise Operations
a = np.array([1, 2, 3])
b = np.array([2, 3, 4])
# Addition
c = a + b        
#  array([3, 5, 7])
# Multiplication
c = a * b        
# Exponent
c = a ** 2       
# array([2, 6, 12])
# array([1, 4, 9])
# Universal Functions
np.sin(a), np.exp(b)
```
